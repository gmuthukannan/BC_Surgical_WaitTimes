```python
#Combine the rows of Names 'Langley, City of' and 'Langley, Township of' into 'Langley' in df_Comb_pop
df_Comb_pop['Name'] = df_Comb_pop['Name'].replace({ 'Langley, City of' : 'Langley', 'Langley, District Municipality' : 'Langley' })
df_Comb_pop['Name'] = df_Comb_pop['Name'].replace({ 'North Vancouver, City of' : 'North Vancouver', 'North Vancouver, District Municipality' : 'North Vancouver' })
df_Comb_pop['Name'] = df_Comb_pop['Name'].replace({
    'Sechelt Ind Gov Dist (Part-Sunshine Coast)' : 'Sechelt District Municipality',
    'Sechelt Ind Gov Dist (Part-qathet)':'Sechelt District Municipality'
})
```

```python
print(df_Comb_pop['Name'].unique())
```

```python
# Group by 'Name' and 'Area Type', and sum all other numeric columns.
df_combPop_aggregated = (
    df_Comb_pop
    .groupby(['Name', 'Area Type'])
    .sum()
    .reset_index()
)
```

```python
#Rename the column 'Name' to 'City' in df_combPop_aggregated so that We can merge with surgical wait time data
df_combPop_aggregated = df_combPop_aggregated.rename(columns={'Name': 'City'})


df_combPop_aggregated.iloc[118:140]
```

```python
year_columns = [col for col in df_combPop_aggregated.columns if str(col).isdigit() and 2001 <= int(col) <= 2024]

# 2. MELT the population DataFrame to long format
df_pop_long = pd.melt(
    df_combPop_aggregated,
    id_vars=['City'],                          # Keep 'City' as the main identifier
    value_vars=year_columns,                   # These are the columns to melt (the population years)
    var_name='Calendar_Year',                  # Name the new column holding the year values
    value_name='Population'                    # Name the new column holding the population counts
)

# 3. Ensure 'Calendar_Year' is an integer to match the surgical data
df_pop_long['Calendar_Year'] = pd.to_numeric(df_pop_long['Calendar_Year'], errors='coerce').astype('Int64')

print(df_pop_long.head())
```

```python
#Merge the Population Data with Surgical Wait Time Data on 'City' column left join so that we retain all surgical wait time records hospital wise and get population data where available
df_Popltn_SurgWt_combined = pd.merge(df_surgical_waitTime_Cleaned, df_pop_long, on=['City','Calendar_Year'], how='left')
```
